```python
import matplotlib.pyplot as plt

import os

import numpy as np
import nibabel as nib
```

```python
pth = 'data\keras_slices_data\keras_slices_seg_test'
img_paths = os.listdir(pth)

crt = []

for img_path in img_paths:
    image = np.array(nib.load(os.path.join(pth, img_path)).dataobj)
    image = np.expand_dims(image, axis=0)
    image = np.moveaxis(image, 0, 0)
    for x in np.unique(image):
        if not x in crt:
            crt.append(x)
            print(crt)
```
